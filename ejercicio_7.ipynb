### a) Verificación de la ortogonalidad del residuo

Tomamos las mediciones correspondientes a $\sigma=0{,}06$ y calculamos su reconstrucción por mínimos cuadrados. Luego obtenemos el residuo:

$$
r_\sigma=b_\sigma-A_7\widehat{x}_\sigma.
$$

Para verificar numéricamente que el residuo es ortogonal a las columnas de $A_7$, calculamos:

$$
A_7^T r_\sigma.
$$

Si su norma es cercana a cero, consideramos verificada la condición.

```python
import numpy as np

from funciones import N, xfantoma, matriz_proyeccion

# Construir la matriz A7
D2 = [(1, 0), (0, 1)]
D4 = D2 + [(1, 1), (1, -1)]
D7 = D4 + [(2, 1), (1, 2), (3, 1)]

A7, rayos7 = matriz_proyeccion(N, D7)

# Mediciones limpias
b7 = A7 @ xfantoma

np.random.seed(0)
z = np.random.randn(len(b7))

escala = np.sqrt(np.mean(b7**2))

#usamos el  ruido de 0.06
sigma = 0.06
eta_sigma = sigma * escala * z
b_sigma = b7 + eta_sigma

x_sombrero_sigma, _, _, _ = np.linalg.lstsq(A7, b_sigma, rcond=None)

r_sigma = b_sigma - A7 @ x_sombrero_sigma

# A7 transpuesta por el residuo
At_r = A7.T @ r_sigma

# Verificar que sea cercano a cero
print("Norma del residuo:", np.linalg.norm(r_sigma))
print("Norma de A7^T @ r_sigma:", np.linalg.norm(At_r))
```

### b)
El producto $A_7^T r_\sigma$ contiene los productos internos entre el residuo y cada columna de $A_7$. Por lo tanto, al obtener

$$
A_7^T r_\sigma \approx 0,
$$

estamos verificando que el residuo es aproximadamente ortogonal a todas las columnas de $A_7$. Como el espacio columna $\operatorname{Col}(A_7)$ está formado por todas las combinaciones lineales de esas columnas, el residuo también es ortogonal a todo ese espacio.

Que el residuo sea ortogonal a las columnas de $A_7$ significa que la parte de las mediciones que quedó sin reproducir no puede reducirse modificando los píxeles de la reconstrucción. Por lo tanto, $\widehat{x}_\sigma$ es la solución que mejor aproxima las mediciones ruidosas en el sentido de mínimos cuadrados.

Por otro lado, el vector

El vector $A_7\widehat{x}_\sigma$ representa las mediciones que produciría la imagen reconstruida. Este vector pertenece a $\operatorname{Col}(A_7)$ porque es una combinación lineal de las columnas de $A_7$.

Debido al ruido, las mediciones $b_\sigma$ no pueden reproducirse exactamente. Por eso, mínimos cuadrados busca, dentro de $\operatorname{Col}(A_7)$, el vector más cercano a $b_\sigma$. Ese vector es $A_7\widehat{x}_\sigma$.

La diferencia entre ambos es el residuo:

$$
r_\sigma=b_\sigma-A_7\widehat{x}_\sigma.
$$

Como el residuo es ortogonal a $\operatorname{Col}(A_7)$, decimos que $A_7\widehat{x}_\sigma$ es la proyección ortogonal de $b_\sigma$ sobre ese espacio.

Las proyecciones tomográficas son las mediciones realizadas por el escáner desde distintas direcciones. Cada medición representa la atenuación acumulada a lo largo de un rayo. En este ejercicio, estas mediciones están representadas por el vector $b_\sigma$.

En cambio, la proyección ortogonal aparece al resolver el problema de mínimos cuadrados. Como las mediciones contienen ruido, generalmente $b_\sigma$ no pertenece a $\operatorname{Col}(A_7)$ y no existe una imagen que pueda reproducirlo exactamente.

Por eso, mínimos cuadrados busca el vector de mediciones posibles más cercano a $b_\sigma$. Este vector es $A_7\widehat{x}_\sigma$ y se lo denomina proyección ortogonal de $b_\sigma$ sobre $\operatorname{Col}(A_7)$.

En resumen, las proyecciones tomográficas son mediciones físicas tomadas por el escáner, mientras que la proyección ortogonal es la aproximación de las mediciones ruidosas mediante un vector que sí puede ser producido por el modelo.
